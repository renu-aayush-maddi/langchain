```python
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
model = init_chat_model("google_genai:gemini-2.5-flash-lite")
model
```

```python
from pydantic import BaseModel, Field

class Movie(BaseModel):
    title: str = Field(description="The title of the movie")
    director: str = Field(description="The director of the movie")
    release_year: int = Field(description="The year the movie was released")
    rating: float = Field(description="The rating of the movie out of 10")
```

```python
model_with_structured_output = model.with_structured_output(Movie)
model_with_structured_output
```

```python
model_with_structured_output.invoke("What is the movie Inception about?")
```

```python
from pydantic import BaseModel, Field

class Actor(BaseModel):
    name: str = Field(description="The name of the actor")
    role: str = Field(description="The role of the actor in the movie")
    

class MovieDetails(BaseModel):
    title: str = Field(description="The title of the movie")
    director: str = Field(description="The director of the movie")
    release_year: int = Field(description="The year the movie was released")
    rating: float = Field(description="The rating of the movie out of 10")
    actors: list[Actor] = Field(description="A list of actors in the movie")
    
model_with_nested_structured_output = model.with_structured_output(MovieDetails)
model_with_nested_structured_output.invoke("What is the movie Inception about?")
```

```python
#TypedDict
#Ideal when you dont need runtime validation and want to use type hints for static analysis. It is a more lightweight option compared to Pydantic models, which provide runtime validation and additional features. TypedDict is suitable for simple use cases where you just want to define the structure of a dictionary without the overhead of a full-fledged model.
```

```python
from typing_extensions import TypedDict,Annotated

class MovieDict(TypedDict):
    title: Annotated[str, "The title of the movie"]
    director: Annotated[str, "The director of the movie"]
    release_year: Annotated[int, "The year the movie was released"]
    rating: Annotated[float, "The rating of the movie out of 10"]
    
model_with_typeddict_output = model.with_structured_output(MovieDict)
response = model_with_typeddict_output.invoke("Return ONLY a valid movie object with title, director, release_year, and rating for the movie Inception.")
print(response)
```

```python
# Structured output using a Pydantic schema
# Pydantic models are the supported approach for structured output in this LangChain version.
```

```python
from pydantic import BaseModel, Field
from langchain.chat_models import ChatOpenAI

# Define model
model = ChatOpenAI(model="gpt-4.1-mini")

# Define structured schema
class ContactInfo(BaseModel):
    name: str = Field(description="The name of the contact")
    email: str = Field(description="The email address of the contact")
    phone: str = Field(description="The phone number of the contact")

# Create structured model
structured_model = model.with_structured_output(ContactInfo)

# Invoke model
result = structured_model.invoke(
    "Extract contact information from the following text: "
    "'John Doe, email: john.doe@example.com, phone: 123-456-7890'"
)

# Print structured output
print(result)
```
